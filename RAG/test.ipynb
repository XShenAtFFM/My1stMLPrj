```python
import pymupdf
```

```python
doc = pymupdf.open("./documents/004.pdf") # open a document
```

```python
out = open("001.txt", "wb") # create a text output
for page in doc: # iterate the document pages
    text = page.get_text().encode("utf8") # get plain text (is in UTF-8)
    out.write(text) # write text of page
    out.write(bytes((12,))) # write page delimiter (form feed 0x0C)
out.close()
```

```python
doc = pymupdf.open("./documents/002.pdf") # open a document
```

```python
with open("004_clean.txt", "w", encoding="utf-8") as out:
    for page in doc:
        blocks = page.get_text("blocks")
        
        # sort blocks by vertical position (top to bottom)
        blocks = sorted(blocks, key=lambda b: (b[1], b[0]))
        
        for block in blocks:
            text = block[4]
            out.write(text)
            out.write("\n")
        
        out.write("\n\n=== PAGE BREAK ===\n\n")
```

```python
with open("004_clean_dict.txt", "w", encoding="utf-8") as out:
    for page in doc:
        page_dict = page.get_text("dict")
        for block in page_dict["blocks"]:
            if block["type"] == 0:  # text block
                for line in block["lines"]:
                    for span in line["spans"]:
                        out.write(span["text"])
                    out.write("\n")
```

```python
import re
```

```python
with open("./chunks/002.txt", "r") as f:
    text = f.read()
#chapters = re.split(r'\n(\d+(?:\.\d+){0,2})\s*\n', text)
chapters = re.split(r'\n(\d+(?:\.\d+)?(?:\.\d+)?)\s*\n', text)
chunks02 = []
chapter_idx = ''
chapter_pairs = zip(chapters[1::2], chapters[2::2])
#chapter_idx_pattern = re.compile('\d+(?:\.\d+){0,2}\s*')
chapter_idx_pattern = re.compile('\d+(?:\.\d+)?(?:\.\d+)?\s*')
for chapter_idx, chapter in chapter_pairs:
    #print(chapter_idx)
    if chapter.count('\n') < 2:
        continue    
    assert(chapter_idx_pattern.match(chapter_idx))
    
    words = chapter.split()
    for i in range(0, len(words), chunk_size - overlap):
        chunk_words = words[i : i + chunk_size]
        chunk_text = " ".join(chunk_words)
        chunk_id = "chunk_" + str(len(chunks02) +1 ).zfill(5)
        chunk = {'id' : chunk_id, "source" : "file_name", "chapter" : chapter_idx, "text" : chunk_text}
        chunks02.append(chunk)
```
